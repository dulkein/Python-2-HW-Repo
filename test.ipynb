```python
import time

class Timer:
    def __enter__(self):
        self.start = time.time() 
        return self                     
       
    def __exit__(self, exc_type, exc_value, traceback):
        self.end = time.time()   
        total = self.end - self.start
        print(f"Total time: {total:.2f} ms")

def fibo(n):
    if n <= 1:
        return n
    return fibo(n-1) + fibo(n-2)

with Timer() as t:
    fibo(35)
```
